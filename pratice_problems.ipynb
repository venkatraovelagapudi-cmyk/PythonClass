```python
#3. find reverse of a number
def rever_of_digits(num):
    rev_no=0
    if num<0:
        sign=-1
        num=abs(num)
    else:
        sign=1
    while num>0:
        ls_di=num%10
        rev_no= rev_no*10+ls_di
        num=num//10
    return rev_no*sign
print(rever_of_digits(1234))
print(rever_of_digits(-91234))
```

```python
# 5.check weather number is armstrogn number
def amstrong_no(num):
    temp=num
    sum=0
    while num>0:
        ls_di=num%10
        sum= sum+ls_di**len(str(temp))
        num=num//10
    if temp==sum:
        return "amstrong"
    else:
        return "Not amstrong"
print(amstrong_no(153))
print(amstrong_no(9))   
print(amstrong_no(183))
```

```python
#SUM OF THE DIGITS USING RECURSION
def sum_of_digits(n,res=0):
    if n==0:
        return res
    return sum_of_digits(n//10,res+n%10)
print(sum_of_digits(1234))
print(sum_of_digits(153))
```
