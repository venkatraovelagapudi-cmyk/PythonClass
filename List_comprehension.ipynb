```python
#printing numbers from 1to n
n=int(input())
lst=[i for i in range(1,n+1)]
print(lst)
```

```python
#printing numbers from 1to n and only even
n=int(input())
lst=[i for i in range(1,n+1) if i&1==0]
print(lst)
```

```python
#printing numbers from 1to n and only odd and bitwise and
n=int(input())
lst=[i for i in range(1,n+1) if i&1]
print(lst)
```

```python
# add 1 if number is even otherwise and 0.5 for ist of numbers
lst=[1,2,3,4,5,7]
res=[num+1 if num%2==0 else num+0.5 for num in lst]
print(res)
```

```python
n=6
res=[(i,j) for i in range(n+1) for j in range(n+1) ]
print(res)
```

```python
# return combation where sum is equal to 12
n=6
res=[(i,j) for i in range(n+1) for j in range(n+1) if i+j==12 ]
print(res)
```

```python
# return all combinations if any dice face is 4
n=6
res=[(i,j) for i in range(n+1) for j in range(n+1) if i==4 or j==4 ]
print(res)
```

```python
# return even if sum of two dices is even or odd
n=6
res=[f"{i}+{j}:EVEN"if (i+j)%2==0 else f"{i}+{j}:ODD" for i in range(n+1) for j in range(n+1) ]
print(res)
```
