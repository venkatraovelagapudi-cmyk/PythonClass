```python
def seq(x):
    a=x+5
    yield a
    a+=2
    yield x+a
    yield x+2
res=seq(5)
print(res)
# print(next(res))
# print(next(res))
# print(next(res))
for num in res:
    print(num)
# print(next(res))#StopIteration: 
```

```python
def seq(n):
    res=[]
    for num in range(n+1):
        yield num
res=seq(11)
for i in res:
    print(i)
```
