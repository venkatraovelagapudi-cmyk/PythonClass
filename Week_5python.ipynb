```python
def calculate(a,b,op='add'):
    if op=='add':
        return a+b
    elif op=='mul':
        return a*b
    else:
        return 0
print(calculate(3,4))
print(calculate(3,4,'mul'))
```

```python
def is_palimdrome_num(n):
    temp=n
    rev=0
    while n>0:
        last_digit=n%10
        rev=rev*10+last_digit
        n=n//10
    if temp==rev:
        return True
    else:
        return False
print(is_palimdrome_num(121))
print(is_palimdrome_num(1331))
print(is_palimdrome_num(12321))
print(is_palimdrome_num(-121))
print(is_palimdrome_num(123))
```

```python
def is_palindrome_str(s):
    palindrome=''.join([i.lower() for i in s if i.isalpha()])
    #print(palindrome)
    if palindrome==palindrome[::-1]:
        return True
    else:
        return False
print(is_palindrome_str("racecar"))
print(is_palindrome_str("A man a plan a canal Panama"))
print(is_palindrome_str("hello"))
```

```python
di={}
def operation_on_list(lst):
    len_lst=len(lst)
    sum_lst=sum(lst)
    di['min']=min(lst)
    di['max']=max(lst)
    di['mean']=sum_lst/len_lst
    di['sum']=sum_lst
    return di
print(operation_on_list([4,8,15,16,23,42]))
    
```

```python
gradebook = {}
def add_student(gradebook, name, marks_dict):
    gradebook[name] = marks_dict
add_student(gradebook, 'Alice', {'Math':85, 'Science':90, 'English':78})
add_student(gradebook, 'Bob', {'Math':55, 'Science':62, 'English':70})
add_student(gradebook, 'Carol', {'Math':92, 'Science':88, 'English':95})
def top_student(gradebook):
    top_name = None
    max_avg = 0

    for student, marks in gradebook.items():
        avg = sum(marks.values()) / len(marks)

        if avg > max_avg:
            max_avg = avg
            top_name = student

    return top_name
print(top_student(gradebook))
def pass_fail_list(gradebook, pass_mark=40):
    passed = []
    failed = []

    for student, marks in gradebook.items():
        if all(mark >= pass_mark for mark in marks.values()):
            passed.append(student)
        else:
            failed.append(student)

    return passed, failed
print(pass_fail_list(gradebook))
```
